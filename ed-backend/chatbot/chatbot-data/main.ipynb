```python
import re
from pathlib import Path
from langchain_community.document_loaders import TextLoader, DirectoryLoader
from langchain_text_splitters import CharacterTextSplitter, RecursiveCharacterTextSplitter
from langchain_core.documents import Document
```

```python
def load_doc():
    with open("backend/data/test_chunking/neda_ed_men.txt", "r", encoding="utf-8") as f:
        doc = f.read()
    
    return Document(
        page_content=doc
    )
```

```python
def load_docs():
    loader = DirectoryLoader("backend/data/test_chunking", glob="*.txt")
    loaded_txts = loader.load()
    return loaded_txts
```

```python
def chunking(loaded_txts):
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size = 200,
        chunk_overlap = 50,
        separators=["\n", "\n\n", " ", ""]
    )

    chunks = text_splitter.split_documents(loaded_txts)

    print(f"Split {len(loaded_txts)} docs into {len(chunks)} chunks")
    #print(chunks[:])

    return chunks
```

```python
def main():
    docs = load_docs()
    chunk_txt = chunking(docs)
    return chunk_txt
```

```python
main()
```

```python
texts = main()
print(texts)
```

```python
from langchain_ollama import OllamaEmbeddings


embeddings = OllamaEmbeddings(model="llama3")
```

```python
documents_list = [doc.page_content for doc in texts]
vector = embeddings.embed_documents(texts=documents_list)
print(vector)
```

```python
collection = chroma_client.get_or_create_collection(name="embeddings_test")
```

```python
collection.add(
    ids=[f"mens_health_food_insecurity{i}" for i in range(len(texts))],
    embeddings=vector,
    documents=documents_list,
    metadatas=[{"source": "neda_ed_men.txt"} for _ in range(len(texts))]
)
```

```python
question_embedded = embeddings.embed_query("What are some links between food insecurity and eating disorders?")

answer = collection.query(
    query_embeddings=[question_embedded],
    n_results=5
)
```
